# baseline_vs_p4_sequential_batch — results inspection

**Every cell is self-contained.** Run any single cell on its own — it re-imports `nb_plot` and renders its plot without needing the cells above it.

- `nb_plot.single_run(N)` — one run, all three methods
- `nb_plot.all_runs(METHOD)` — every run overlaid for one method (+ mean)
- `nb_plot.compare(M1, M2, ...)` — aggregate mean ± std across runs
